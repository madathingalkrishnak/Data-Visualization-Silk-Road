```python
from pyngrok import ngrok
import subprocess
import time

# Kill any existing Streamlit processes
ngrok.kill() 
!pkill -f streamlit

# Start fresh ngrok tunnel
public_url = ngrok.connect(addr="8501", proto="http")
print("🌐 Streamlit app URL:", public_url)

# Run Streamlit using subprocess
process = subprocess.Popen(["streamlit", "run", "streamlit_appp.py"])

# Wait for the app to start
time.sleep(5)

# Check if the Streamlit process started correctly
if process.poll() is not None:
    print("❌ Streamlit process failed to start!")
    out, err = process.communicate()
    print("Error output:", err)
else:
    print("✅ Streamlit is running in the background.")
```
