```python
import pandas as pd
import numpy as np
import random
import plotly.express as px
import plotly.graph_objects as go
import streamlit as st

# Load data
df_categories = pd.read_csv('Categories.csv').dropna(axis=0)
df_feedback = pd.read_csv('feedback_.csv')
df_item = pd.read_csv('item.csv').dropna(axis=0)
df_price = pd.read_csv('price.csv')

# Merge DataFrames
df1 = pd.merge(df_item, df_price, on='item_id')
df2 = pd.merge(df1, df_feedback, on='item_id')
df_categories.rename(columns={'Category_id': 'category', 'Category': 'category_name'}, inplace=True)
df_categories['category'] = df_categories['category'].str.split(',')
df_categories = df_categories.explode('category')
df_categories['category'] = df_categories['category'].astype('int64')
df_categories = df_categories.reset_index(drop=True)
df = pd.merge(df2, df_categories, on='category')
```

```python
df.columns
```

```python
# Print unique values for shipping origin and destination
print("Unique Shipping Origins:")
print(df['ships_from'].unique())

print("\nUnique Destinations:")
print(df['ships_to'].unique())
```

```python
df['ships_from'] = df['ships_from'].str.replace('Ships from:', '', regex=False)  # Remove 'Ships from:'
df['ships_from'] = df['ships_from'].str.replace(r'<br />', '', regex=True).str.strip()

df['ships_from'] = df['ships_from'].str.strip()  

df['ships_from'] = df['ships_from'].replace({
    'United States': 'United States of America',
    'United States of America <br />': 'United States of Americas',
    'US': 'United States of America',
    'USA': 'United States of America',
    'us': 'United States of America',
    'UK': 'United Kingdom',
    'germany': 'Germany',
    'The Netherlands': 'Netherlands',
    'worldwide': 'Worldwide',
    'world': 'Worldwide',
    'anywhere': 'Anywhere',
    'Anywhere': 'Anywhere',
    'High Speed DL': 'Direct Download',
    'Direct Download': 'Direct Download',
    'Fast Direct DL': 'Direct Download',
    'www': 'Direct Download',
    'Direct DL': 'Direct Download',
    'internet': 'Direct Download',
    'undeclared': 'Unknown'
})

# Print unique values after cleaning
print("Unique Shipping Origins after cleaning:")
print(df['ships_from'].unique())
```

```python
df['ships_to'] = df['ships_to'].str.replace('Ships from:', '', regex=False)  # Remove 'Ships from:'
df['ships_to'] = df['ships_to'].str.strip() 
df['ships_to'] = df['ships_to'].str.replace(r'<br />', '', regex=True).str.strip() 

df['ships_to'] = df['ships_to'].replace({
    'United States': 'United States of America',
    'United States of America <br />': 'United States of Americas',
    'US': 'United States of America',
    'USA': 'United States of America',
    'us': 'United States of America',
    'UK': 'United Kingdom',
    'germany': 'Germany',
    'The Netherlands': 'Netherlands',
    'worldwide': 'Worldwide',
    'world': 'Worldwide',
    'anywhere': 'Anywhere',
    'Anywhere': 'Anywhere',
    'High Speed DL': 'Direct Download',
    'Direct Download': 'Direct Download',
    'Fast Direct DL': 'Direct Download',
    'www': 'Direct Download',
    'Direct DL': 'Direct Download',
    'internet': 'Direct Download',
    'undeclared': 'Unknown'
})

# Print unique values after cleaning
print("Unique Shipping Origins after cleaning:")
print(df['ships_from'].unique())
```
